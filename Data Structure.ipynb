## Set

```python
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)
```

### Explanation
In output all the input elements are not visible.It remove duplicated values such as a,i. which means that set does not allow duplicated values and it automatically removes duplicated values.
Also the output order varies from inout,which means that sets are unorderd collections.


```python
my_set[4]='s'
print(my_set)
```

### Explanation
Sets are unorderd collections of data structure and they do not support indexing,so we cannont access or modify 
set elements using indexing position.Thats why my_set[4]='s' produce a TypeError.

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print(set1)
print(set2)
```

```python
print("Union:",set1.union(set2))
```

```python
print("Intersection:",set1.intersection(set2))
```

## Operators & Conditional Statements :

```python
score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent performance! Keep up the great work.")
elif score >= 4 and score <= 7:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
